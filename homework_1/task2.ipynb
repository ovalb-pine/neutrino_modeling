```python
import numpy as np
```

```python
class Source:
    def __init__(self, Y=10000):
        self.Y = Y

    def generate_directions(self, E):
        """Разыгрывает начальные направления фотонов для энергии E (МэВ)."""

        N = int(self.Y * E)

        phi = np.random.uniform(0, 2 * np.pi, N)
        cos_theta = np.random.uniform(-1, 1, N)
        sin_theta = np.sqrt(1 - cos_theta**2)

        directions = np.column_stack((
            sin_theta * np.cos(phi),
            sin_theta * np.sin(phi),
            cos_theta
        ))

        return directions
```

```python
source = Source()

directions = source.generate_directions(E=1)

print("Количество фотонов:", len(directions))
print("Первые 5 направлений:")
print(directions[:5])
```

```python
# проверка нормировки
norms = np.linalg.norm(directions, axis=1)

print(np.allclose(norms, 1))
```
